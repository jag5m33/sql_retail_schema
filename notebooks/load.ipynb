```python
import urllib.request, zipfile, os

print("working dir:", os.getcwd())   # check where you are

os.makedirs('../data', exist_ok=True)

url = 'https://archive.ics.uci.edu/static/public/502/online+retail+ii.zip'
zip_path = '../data/online_retail_II.zip'

urllib.request.urlretrieve(url, zip_path)

with zipfile.ZipFile(zip_path) as z:
    print(z.namelist())
    z.extractall('../data')

print(os.listdir('../data'))
```

```python
!pip install openpyxl
import pandas as pd

path = '../data/online_retail_II.xlsx'
print(pd.ExcelFile(path).sheet_names)

df1 = pd.read_excel(path, sheet_name='Year 2009-2010')
df2 = pd.read_excel(path, sheet_name='Year 2010-2011')
df = pd.concat([df1, df2], ignore_index=True)

print(df.shape)
df.head()
```

```python
os.remove('../data/online_retail_II.zip')
```

```python
import sqlite3

df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
print(df.columns.tolist())

conn = sqlite3.connect('../data/retail.db')
df.to_sql('flat', conn, if_exists='replace', index=False)
print(pd.read_sql('SELECT COUNT(*) AS n FROM flat', conn))
conn.close()
```
